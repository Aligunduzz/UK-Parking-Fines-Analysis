```python
df.plot(x="Month", y="Penalty Charge Notices Issued")
```

```python
df.plot(x="Month", y="On & Off Street Income from PCNs")
```

```python
fig, ax1 = plt.subplots()

# PCN sayısı
ax1.plot(
    df["Month"],
    df["Penalty Charge Notices Issued"],
    label="Penalty Charge Notices",
    color="blue"
)

ax1.set_xlabel("Month")
ax1.set_ylabel("PCNs Issued", color="blue")
ax1.tick_params(axis="y", labelcolor="blue")


# Gelir
ax2 = ax1.twinx()

ax2.plot(
    df["Month"],
    df["On & Off Street Income from PCNs"],
    label="PCN Income",
    color="red"
)

ax2.set_ylabel("Income (£)", color="red")
ax2.tick_params(axis="y", labelcolor="red")

plt.title("Monthly PCNs Issued and Income")
plt.show()
```
